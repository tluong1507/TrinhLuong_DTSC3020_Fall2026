# Object-Oriented Programming (OOP) in Python

Today, we are going to learn how to organize Python code using **classes and objects**.

# 1. What is Object-Oriented Programming?

Imagine we are building a program for a university.
For each student, we need to store a name, a major, and a number of credits.

We also want our program to add credits, check if a student is full-time, and show student information.

OOP helps us keep this related information and these actions together.
This makes our code easier to organize and reuse.

Object-Oriented Programming, or **OOP**, is a way to organize a program.

In OOP, we group two things together:

- **data**
- **actions that work with that data**

Instead of keeping many separate variables and functions, we can put related information and behavior together inside a **class**.




# 2. Class, Object, Attributes, and Methods

- A **class** is a **blueprint**.

- An **object** is a real thing made from **that blueprint**.

- Think of a **class** as a **cookie cutter**, and an object as an actual cookie made from it.

- **Attributes** are the object's **DATA**.
  - Example: `shape` can be round, star, or heart.
  - Example: `flavor` can be chocolate or vanilla.

- **Methods** are actions we can ask the object to perform.
  - Example: a cookie can `describe()` itself.

- We combine attributes (**data**) and methods (**behaviors**) in one place.

- Each object can have its own data.
  - `cookie1` → star, chocolate
  - `cookie2` → heart, vanilla


# 3. Understanding `__init__()` and `self`

- `__init__()` is a **special method** in a class.

- Python calls it automatically when we create a new object.

- We use it to give the object its **starting data**.

- For a cookie, the starting data can be:
  - `shape`
  - `flavor`

- `self` means **this particular object**.

When we write:

```python
self.shape = shape
self.flavor = flavor
```

we are storing the shape and flavor inside **this particular cookie object**.

### Simple way to remember

- `__init__()` → gives the object its starting information.
- `self` → means **this object**.


# 4. Example 1 — Basic Class and Object: `Cookie`

**Concept focus:** class, object, attributes, methods, `__init__()`, and `self`.


````python
class Cookie:
                 #self refers to the object that calls the method later(at object creation time).

    def __init__(self, shape, flavor):  # we use it to give the object its starrting information.
                                        # __init__ (dunder init) runs automatically when we create a Cookie later.


        self.shape = shape      # Store this shape inside this particular cookie.
        self.flavor = flavor    # Store this flavor inside this particular cookie.



    def describe(self):   # 'describe' is a method (an action objects can do).
        print(f"This is a {self.shape}-shaped cookie with {self.flavor} flavor.")


# create two Cookie objects from the Cookie class:
cookie1 = Cookie("star", "chocolate")
  #cookie1.shape = "star"
  #cookie1.flavor = "chocolate"
cookie2 = Cookie("heart", "vanilla")

# Call the 'describe' method on each object to see their data/behavior in action:
cookie1.describe()
cookie2.describe()
````

````python
class Cookie:

    def __init__(self, shape, flavor):
        self.shape = shape
        self.flavor = flavor

    def describe(self):
        print(f"This is a {self.shape}-shaped cookie with {self.flavor} flavor.")


cookie1 = Cookie("star", "chocolate")
cookie2 = Cookie("heart", "vanilla")

cookie1.describe()
cookie2.describe()
````

# 5. Example 2 — Working with Methods: `Student`

We want to create a `Student` class with **name, major, and credits**.

The student should be able to:

- add credits
- check full-time status
- show a summary

**Concept focus:** methods can read and change an object's data.


````python
class Student:

    def __init__(self, name, major, credits=0):
        self.name = name
        self.major = major
        self.credits = credits

    def add_credits(self, amount):
        # Add credits to this student
        self.credits = self.credits + amount
        # self.credits += amount


    def is_full_time(self):
        # A student is full-time with 12 or more credits
        if self.credits >= 12:
            return True
        else:
            return False

    def summary(self):
        # Decide the student's status
        if self.credits >= 12:
            status = "Full-time"
        else:
            status = "Part-time"

        return f"{self.name} - {self.major} ({self.credits} credits, {status})"


# Create Student objects
s1 = Student("Sarah", "Data Science", 9)
s2 = Student("Bob", "Computer Science", 13)

print(s1.summary())
print(s2.summary())

# Change Sarah's data by using a method
s1.add_credits(4)

print(s1.summary())
````

# 6. Example 3 — Inheritance: `Student` → `GraduateStudent`

**Inheritance** means one class can reuse attributes and methods from another class.

- `Student` → **parent class**
- `GraduateStudent` → **child class / subclass**

A `GraduateStudent` **is a** `Student`.

The child class can reuse the parent's code and add new information of its own.


````python
# Parent class
class Student:

    def __init__(self, name, major):
        self.name = name
        self.major = major

    def introduce(self):
        return f"My name is {self.name}, and my major is {self.major}."


# Child class: GraduateStudent inherits from Student
class GraduateStudent(Student):

    def __init__(self, name, major, advisor):
        # Use the parent class to set name and major
        super().__init__(name, major) # super() lets us use code from the parent class.

        # Add information specific to GraduateStudent
        self.advisor = advisor
# GraduateStudent inherits the attributes and methods of Student, and then adds its own new information.

student1 = GraduateStudent("Sarah", "Data Science", "Dr. Smith")

# introduce() is inherited from Student
print(student1.introduce())

# advisor belongs to GraduateStudent
print(f"Advisor: {student1.advisor}")
````

### Key Idea

```python
class GraduateStudent(Student):
```

means `GraduateStudent` inherits from `Student`.

```python
super().__init__(name, major)
```

uses the parent class's `__init__()` to set up `name` and `major`.

**Inheritance = IS-A relationship**

> A GraduateStudent **is a** Student.


# 7. Example 4 — Composition: `Camera` has a `MemoryCard`

**Composition** means one object contains or uses another object.

A `Camera` **has a** `MemoryCard`.

**Composition = HAS-A relationship**


````python
class MemoryCard:

    def __init__(self, capacity):
        self.capacity = capacity   # Total space
        self.used = 0              # Used space starts at 0

    def free_space(self):
        # Calculate remaining space
        return self.capacity - self.used

    def write(self, size):
        # Add data to the memory card
        self.used += size

    def delete(self, size):
        # Remove data from the memory card
        self.used -= size


class Camera:

    def __init__(self, model, card):
        self.model = model
        self.card = card   # Store a MemoryCard object inside the Camera

    def take_photo(self, size):
        # Use the MemoryCard object's write() method
        self.card.write(size)

    def status(self):
        print("Used space:", self.card.used)
        print("Free space:", self.card.free_space())


# Create a MemoryCard object
card1 = MemoryCard(100)

# Create a Camera object and give it the memory card
camera1 = Camera("Photon X", card1)

# Take photos
camera1.take_photo(20)
camera1.take_photo(10)

camera1.status()

# Delete some data from the card
camera1.card.delete(5)

camera1.status()
````

### Inheritance vs. Composition

- **Inheritance:** `GraduateStudent` **is a** `Student`.
- **Composition:** `Camera` **has a** `MemoryCard`.

This distinction helps us decide how classes should be related.


# 8. Exercise — `Rectangle`

Create a simple `Rectangle` class.

Each rectangle should have:

- `width`
- `height`

Add methods to calculate:

- area
- perimeter


````python

class Rectangle:
    def __init__(self, width, height):
        self.width = float(width)    # store the rectangle's width
        self.height = float(height) # store the rectangle's height

    def area(self):
        # multiply width by height to get total area
        return self.width * self.height

    def perimeter(self):
        # add both sides (width + height) and double it
        return 2 * (self.width + self.height)


# create a rectangle with width 3 and height 5
r = Rectangle(3, 5)
print(r.area(), r.perimeter())
````
