## GenAI-Era Example 1: Verify an AI Claim

Suppose an AI tool says:

> If a student studies 7 hours per week for 8 weeks, the total is 54 hours.

Use Python to verify this.


````python
hours_per_week = 7
weeks = 8

total_hours = hours_per_week * weeks

print("Correct total hours:", total_hours)
````

## GenAI-Era Example 2: Detect a Logic Bug

A GenAI tool may generate code that works for normal cases but fails for edge cases.

The following code calculates an average.


````python
scores = [90, 80, 100]
average = sum(scores) / len(scores)
print("Average:", average)
````

Now try an empty list. This is an **edge case**.


````python
scores = []

if len(scores) == 0:
    print("No scores available. Cannot calculate average.")
else:
    average = sum(scores) / len(scores)
    print("Average:", average)
````

## GenAI-Era Example 3: Check Whether a Response Meets Requirements

Suppose you ask an AI tool to write a short course description that must mention Python, data science, and AI. Python can check whether required terms appear.


````python
ai_generated_text = '''
This course introduces Python programming for data science, automation, and AI applications.
Students will learn to write, test, and improve code.
'''

required_terms = ["Python", "data science", "AI"]

for term in required_terms:
    appears = term.lower() in ai_generated_text.lower()
    print(f"{term}: {appears}")
````

## GenAI-Era Example 4: Clean Messy Text

AI-generated or human-written text often contains inconsistent spacing and capitalization. Python can clean it.


````python
raw_topics = [" python ", "PYTHON", "Data Science ", " data science", "AI", " ai "]

cleaned_topics = []

for topic in raw_topics:
    cleaned_topics.append(topic.strip().lower())

print(cleaned_topics)
````

## Preview: Class

````python
class Student:
    def __init__(self, name, major):
        self.name = name
        self.major = major

    def introduce(self):
        print(f"My name is {self.name}, and my major is {self.major}.")

student = Student("Alice", "Data Science")
student.introduce()
````

## Beginner-Friendly Best Practices

| Best practice | What it means in this course |
|---|---|
| Use clear names | Use `student_score`, not `x`, when possible. |
| Write comments | Explain non-obvious logic. |
| Use GitHub | Submit clean repositories with `.py` or `.ipynb` files. |
| Fix broken code early | Do not build on top of errors. |
| Use readable formatting | Use spaces, indentation, and clear structure. |
| Avoid magic numbers | Store important values in named variables. |
| Test with examples | Do not assume code works after one run. |
