# Part 1. Why Control Flow Matters

In Weeks 1–3, we learned values, variables, strings, operators, and data structures. This week, we learn how to make programs behave intelligently.

A real Python program often needs to:

- Make decisions based on conditions
- Repeat work for every item in a list
- Continue until valid input is received
- Choose one action from several possible options
- Avoid crashing when a user enters invalid data
- Recover gracefully when something unexpected happens

Control flow is what turns isolated Python statements into useful applications.


# Part 2. If, If-Else, Elif, and Nested Statements

**Assigned reading:** [Python Decision Making Statements — DataFlair](https://data-flair.training/blogs/python-decision-making-expressions/)

Decision-making statements allow a program to choose which block of code to run based on a condition. DataFlair emphasizes `if`, `if-else`, `elif`, nested `if`, and single-statement conditions.


<div class="figure"><p class="figcap">Comparison of <code>if</code>, <code>if-else</code>, and nested <code>if</code> structures.</p></div>

## Basic If Statement

An `if` statement runs a block only when the condition is true.


````python
score = 85

if score >= 70:
    print("The student passed.")
````

## If-Else Statement

Use `else` to define what should happen when the condition is false.


````python
score = 65

if score >= 70:
    print("The student passed.")
else:
    print("The student did not pass.")
````

## Elif Ladder

Use `elif` when checking multiple conditions.


````python
score = 88

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print("Letter grade:", grade)
````

## Nested If Statement

A nested `if` means one `if` statement inside another. It is useful when a second check should only happen after the first check passes.


<div class="figure"><p class="figcap">Workflow of a nested decision process.</p></div>

````python
score = 92
attendance = 0.95

if score >= 90:
    if attendance >= 0.90:
        print("Excellent performance and attendance.")
    else:
        print("Excellent score, but attendance needs improvement.")
else:
    print("Score below excellence threshold.")
````

## Real Application Example: Course Access Decision

Suppose a student can access a bonus assignment only if they submitted the academic honesty statement and completed Assignment 1.


````python
academic_honesty_submitted = True
assignment1_completed = False

if academic_honesty_submitted and assignment1_completed:
    print("Bonus assignment unlocked.")
elif academic_honesty_submitted and not assignment1_completed:
    print("Please complete Assignment 1 first.")
else:
    print("Please submit the academic honesty statement first.")
````

<div class="warning">
<b>Common error:</b> Python requires a colon after an <code>if</code>, <code>elif</code>, or <code>else</code> header, and the code block must be indented consistently.
</div>


# Part 3. For Loops

**Assigned reading:** [Python Loop — DataFlair](https://data-flair.training/blogs/python-loop/)

A `for` loop iterates over a sequence of items. Python uses the `in` keyword to iterate over items.


<div class="figure"><p class="figcap">Comparison of <code>for</code> loops and <code>while</code> loops.</p></div>

<div class="figure"><p class="figcap">Workflow of a <code>for</code> loop: take each item from a sequence, run the loop body, repeat, then exit.</p></div>

## Looping Through a List


````python
students = ["Alice", "Ben", "Carla"]

for student in students:
    print("Welcome,", student)
````

## Looping Through a String


````python
course = "Python"

for letter in course:
    print(letter)
````

## Looping Through a Dictionary


````python
student = {
    "name": "Alice",
    "major": "Data Science",
    "score": 95
}

for key, value in student.items():
    print(key, ":", value)
````

## Real Application Example: Process Student Scores

Use a loop to classify each score.


````python
scores = [95, 82, 67, 74, 58, 90]

for score in scores:
    if score >= 70:
        print(score, "-> passing")
    else:
        print(score, "-> needs support")
````

<div class="figure"><p class="figcap">Comparison of <code>break</code>, <code>continue</code>, and <code>pass</code>.</p></div>

## Loop Control: break and continue

- `break` exits the loop early.
- `continue` skips the rest of the current iteration and moves to the next item.


````python
# break example: stop after finding the first failing score
scores = [95, 82, 67, 74, 58, 90]

for score in scores:
    if score < 70:
        print("First failing score found:", score)
        break
````

````python
# continue example: skip missing values
scores = [95, None, 82, 67, None, 90]

for score in scores:
    if score is None:
        continue
    print("Valid score:", score)
````

## The `pass` Statement

`pass` is a placeholder statement. It does not stop the loop and it does not skip an iteration. It simply means “do nothing here for now.”

This is useful when you are planning logic but have not implemented it yet.


````python
scores = [95, -1, 82]

for score in scores:
    if score < 0:
        pass   # placeholder: we may handle invalid values later
    else:
        print("Valid score:", score)
````

# Part 4. The range() Function

**Assigned reading:** [Python range() Function — DataFlair](https://data-flair.training/blogs/python-range-function/)

`range()` is an in-built function that generates a sequence of integers. It is often used with `for` loops.


<div class="figure"><p class="figcap">Common ways to call <code>range()</code>.</p></div>

## Three Common Forms


````python
print(list(range(5)))        # stop only
print(list(range(2, 6)))     # start, stop
print(list(range(1, 10, 2))) # start, stop, step
````

## Real Application Example: Numbered Output


````python
tasks = ["Read lesson", "Run examples", "Complete assignment"]

for i in range(len(tasks)):
    print(f"Task {i + 1}: {tasks[i]}")
````

## Better Style with enumerate()

When you need both index and value, `enumerate()` is often more readable.


````python
tasks = ["Read lesson", "Run examples", "Complete assignment"]

for i, task in enumerate(tasks, start=1):
    print(f"Task {i}: {task}")
````

# Part 5. While Loops

**Assigned reading:** [Python while Loops — Real Python](https://realpython.com/python-while-loop/)

A `while` loop repeats as long as a condition remains true. While loops are especially useful when the number of iterations is unknown in advance.


<div class="figure"><p class="figcap">Workflow of a <code>while</code> loop: initialize, check condition, run body, update, repeat or exit.</p></div>

## Basic While Loop


````python
count = 1

while count <= 5:
    print("Count:", count)
    count += 1
````

## Real Application Example: Retry Until Valid Input

This example simulates repeated user input using a list of attempts so it runs safely in a notebook.


````python
attempts = ["abc", "-5", "85"]
attempt_index = 0
score = None

while score is None and attempt_index < len(attempts):
    user_input = attempts[attempt_index]
    attempt_index += 1

    print("Trying input:", user_input)

    if user_input.isdigit():
        value = int(user_input)
        if 0 <= value <= 100:
            score = value
        else:
            print("Score must be between 0 and 100.")
    else:
        print("Please enter a number.")

print("Final valid score:", score)
````

## Avoid Infinite Loops

A `while` loop must eventually become false, or it will run forever.


````python
# Safe loop with a clear stopping condition
counter = 0

while counter < 3:
    print("This loop is safe.")
    counter += 1
````

# Part 6. Switch-Case Style Logic

**Assigned reading:** [Python Switch Case Statement — DataFlair](https://data-flair.training/blogs/python-switch-case/)

Traditional switch-case statements are common in languages such as C, C++, and Java. In Python, similar logic can be implemented in several ways.


<div class="figure"><p class="figcap">Common ways to implement switch-case style logic in Python.</p></div>

## Approach 1: If-Elif-Else

This is the most beginner-friendly method.


````python
menu_choice = "2"

if menu_choice == "1":
    print("View grades")
elif menu_choice == "2":
    print("Submit assignment")
elif menu_choice == "3":
    print("Contact instructor")
else:
    print("Invalid choice")
````

## Approach 2: match-case

`match-case` was introduced in Python 3.10. It provides a cleaner case-style syntax.


````python
menu_choice = "2"

match menu_choice:
    case "1":
        print("View grades")
    case "2":
        print("Submit assignment")
    case "3":
        print("Contact instructor")
    case _:
        print("Invalid choice")
````

## Approach 3: Dictionary-Based Dispatch

This is compact for simple lookup tasks.


````python
menu_actions = {
    "1": "View grades",
    "2": "Submit assignment",
    "3": "Contact instructor"
}

menu_choice = "2"
print(menu_actions.get(menu_choice, "Invalid choice"))
````

## Real Application Example: Simple Chatbot Intent Routing


````python
intent = "deadline"

responses = {
    "deadline": "Assignments are due Friday at 11:59 PM.",
    "office_hours": "Please check the syllabus for office hour details.",
    "zoom": "The course meets online via Zoom."
}

print(responses.get(intent, "Sorry, I do not understand that request."))
````

# Part 8. Python Exception Handling

**Assigned reading:** [Python Exception Handling — DataFlair](https://data-flair.training/blogs/python-exception-handling/)

Exception handling allows a program to recover instead of crashing.

Common structure:

```python
try:
    risky_code
except SomeException:
    handle_error
else:
    run_if_no_error
finally:
    always_run_this
```


<div class="figure"><p class="figcap">Exception handling lets programs recover from expected runtime problems.</p></div>

## Basic try-except


````python
a = 10
b = 0

try:
    result = a / b
    print(result)
except ZeroDivisionError:
    print("You cannot divide by zero.")
````

## Handling Multiple Exception Types


````python
user_input = "abc"

try:
    score = int(user_input)
    print("Score:", score)
except ValueError:
    print("Invalid score: please enter a number.")
except Exception as error:
    print("Something else went wrong:", error)
````

## else and finally


````python
user_input = "95"

try:
    score = int(user_input)
except ValueError:
    print("Invalid score.")
else:
    print("Conversion succeeded. Score:", score)
finally:
    print("Validation attempt complete.")
````

<div class="warning">
<b>Best practice:</b> Avoid using a bare <code>except:</code> unless you have a strong reason. Catch specific exceptions when possible so you do not hide important bugs.
</div>


# Part 9. Applied Mini Case Study: Safe Score Processor

This mini case combines:

- Lists
- For loops
- If-elif-else
- Exception handling
- Summary calculation


<div class="figure"><p class="figcap">Week 4 concepts combined in one practical workflow.</p></div>

````python
raw_scores = ["95", "82", "missing", "67", "100", "-5", "88"]

valid_scores = []
invalid_entries = []

for raw_score in raw_scores:
    try:
        score = int(raw_score)

        if 0 <= score <= 100:
            valid_scores.append(score)
        else:
            invalid_entries.append(raw_score)

    except ValueError:
        invalid_entries.append(raw_score)

print("Valid scores:", valid_scores)
print("Invalid entries:", invalid_entries)

if valid_scores:
    average = sum(valid_scores) / len(valid_scores)
    print("Average valid score:", round(average, 2))
else:
    print("No valid scores available.")
````

## Add Letter Grades


````python
valid_scores = [95, 82, 67, 100, 88]

for score in valid_scores:
    if score >= 90:
        letter = "A"
    elif score >= 80:
        letter = "B"
    elif score >= 70:
        letter = "C"
    elif score >= 60:
        letter = "D"
    else:
        letter = "F"

    print(score, "->", letter)
````

# Part 10. GenAI-Era Examples

## Example 1: Validate AI-Generated Code Output

Suppose a GenAI tool generates predicted labels. We need to check whether all labels are allowed.


````python
ai_labels = ["positive", "neutral", "negative", "unknown", "positive"]
allowed_labels = {"positive", "neutral", "negative"}

for label in ai_labels:
    if label not in allowed_labels:
        print("Invalid AI label found:", label)
    else:
        print("Valid label:", label)
````

## Example 2: Retry a Generated Answer Until It Meets a Rule

This simulates checking multiple AI-generated answers.


````python
generated_answers = [
    "This is too short.",
    "Python is a programming language.",
    "Python is a programming language used for data science and automation."
]

required_words = ["Python", "data", "automation"]

for answer in generated_answers:
    print("Checking:", answer)

    passed = True
    for word in required_words:
        if word.lower() not in answer.lower():
            passed = False

    if passed:
        print("Accepted answer:", answer)
        break
    else:
        print("Rejected: missing required content.")
````

## Example 3: Safe Parser for AI-Generated Numeric Data


````python
ai_outputs = ["0.92", "N/A", "0.75", "error", "1.05"]

valid_probabilities = []

for output in ai_outputs:
    try:
        value = float(output)
        if 0 <= value <= 1:
            valid_probabilities.append(value)
        else:
            print("Out-of-range probability:", output)
    except ValueError:
        print("Could not parse:", output)

print("Valid probabilities:", valid_probabilities)
````
