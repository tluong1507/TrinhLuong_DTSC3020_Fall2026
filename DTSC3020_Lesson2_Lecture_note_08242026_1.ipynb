# Part 2. Variables and Data Types

**Assigned reading:** [Python Variables and Data Types — DataFlair](https://data-flair.training/blogs/python-variables-and-data-types/)

A variable is a name that refers to a value. Python creates a variable when a value is assigned to it.
- A variable can have a short name (like x and y) or a more descriptive name (age, carname, total_volume)
- A variable name must start with a letter or the underscore character
- A variable name cannot start with a number
- A variable name can only contain alpha-numeric characters and underscores (A-z, 0-9, and _ )
- Variable names are case-sensitive
- [Python keywords can’t be used as variables](https://runestone.academy/ns/books/published/fopp/SimplePythonData/VariableNamesandKeywords.html)

Variable operations:

- Assigning a value
- Multiple assignment
- Assigning the same value to multiple variables
- Swapping variables
- Deleting variables with `del`
- Checking a value's data type with `type()`


```python
# Basic variable assignment
student_name = "Alice"
student_score = 95
course = "DTSC 3020"

print(student_name)
print(student_score)
print(course)
```

## Multiple Assignment and Swapping Variables


```python
# Multiple assignment
age, city = 21, "Denton"
print(age, city)

# Assign same value to multiple variables
x = y = z = 0
print(x, y, z)

# Swapping variables
color1, color2 = "red", "blue"
color1, color2 = color2, color1
print(color1, color2)
```

## Dynamic Typing

Python is dynamically typed, meaning a variable can refer to values of different types during program execution.


```python
value = 10
print(value, type(value))

value = "Python"
print(value, type(value))

value = 3.14
print(value, type(value))
```

<div class="warning">
<b>Important:</b> Dynamic typing is convenient, but it also means you must test code carefully. AI-generated code may fail if a variable contains a different type than expected.
</div>


# Part 3. Numeric Data Types and Conversion

**Assigned reading:** [Python Numeric Data Types — DataFlair](https://data-flair.training/blogs/python-number-types-conversion/)

Python's main numeric types include:

| Type | Meaning | Example |
|---|---|---|
| `int` | whole number | `42` |
| `float` | decimal number | `3.14` |
| `complex` | real and imaginary parts | `2 + 3j` |

Type conversion, including `int()`, `float()`, and `complex()`. One important detail is that `int(7.7)` truncates to `7` rather than rounding up.


<div class="figure"><p class="figcap">Core Python data types for Week 2.</p></div>

```python
print(type(42))
print(type(3.14))
print(type(2 + 3j))

print(int(7.7))       # truncates, does not round
print(float(7))
print(complex(2, 3))
```

## Binary, Octal, and Hexadecimal Preview

DataFlair's numeric data type tutorial also introduces different number representations.


```python
binary_value = 0b10
octal_value = 0o10
hex_value = 0xF

print(binary_value)   # 2
print(octal_value)    # 8
print(hex_value)      # 15
```

# Part 4. Identifiers

**Assigned reading:** [Identifiers in Python — DataFlair](https://data-flair.training/blogs/identifiers-in-python/)

An identifier is a name used for variables, functions, classes, modules, and objects.

Good identifiers make code easier to read, debug, and review.


## Identifier Rules

A Python identifier:

- Can contain letters, digits, and underscores
- Cannot start with a digit
- Cannot contain spaces
- Cannot use special symbols such as `$` or `-`
- Should not be a Python keyword such as `class`, `for`, or `if`


```python
import keyword

names = ["student_name", "2score", "student name", "total_score", "class", "score$"]

for name in names:
    print(name, "| identifier?", name.isidentifier(), "| keyword?", keyword.iskeyword(name))
```

## Naming Practice

Use meaningful names. In this course, prefer `snake_case` for variables and functions.


```python
# Less clear
x = 95

# More clear
assignment_score = 95

print(assignment_score)
```

# Part 5. Variable Scope

**Assigned reading:** [Python Variable Scope — DataFlair](https://data-flair.training/blogs/python-variable-scope/)

Scope means where a variable can be seen and used.

DataFlair explains four important scope ideas:

| Scope | Meaning |
|---|---|
| Local | Variable created inside a function |
| Global | Variable created outside functions |
| Built-in | Names Python already provides, such as `print`, `len`, and `sum` |
| Enclosed | Variable in an outer function used by an inner function |


```python
course = "DTSC 3020"   # global variable

def show_course():
    print("Inside function:", course)

show_course()
print("Outside function:", course)
```

## Local Scope


```python
bonus = 5
def calculate_bonus():
    # bonus = 5   # local variable
    print("Inside function, bonus =", bonus)

calculate_bonus()

# Uncommenting the next line will cause an error because bonus is local.
print(bonus)
```

## Enclosed Scope Preview


```python
def outer_function():
    message = "Hello from outer function"

    def inner_function():
        print(message)

    inner_function()

outer_function()
```

# Part 6. Strings

**Assigned reading:** [Python String — DataFlair](https://data-flair.training/blogs/python-string/)

A string is a sequence of characters. DataFlair covers string quotes, multiline strings, immutability, indexing, slicing, escape sequences, functions, and operations.


<div class="figure"></div>

## Quotes and Multiline Strings


```python
single_quote = 'Dogs are "love"'
double_quote = "Python is 'fun'"
multi_line = """Hello
Welcome to DTSC 3020"""

print(single_quote)
print(double_quote)
print(multi_line)
```

## Indexing, Slicing, and Immutability

Strings can be accessed by index, but they cannot be changed in place.


```python
word = "Python"

print(word[0])
print(word[-1])
print(word[0:3])

# Strings are immutable:
# word[0] = "J"   # This would cause an error

new_word = "J" + word[1:]
print(new_word)
```

## Escape Sequences and String Functions


```python
print("Hello\nWorld")
print("Name:\tAlice")

text = "  Python Programming  "
print(len(text))
print(text.lower())
print(text.upper())
print(text.strip())
print(text.replace("Python", "AI"))

colors = ["red", "green", "blue"]
print("*".join(colors))
```

# Part 7. Operators

**Assigned reading:** [Python Operator — DataFlair](https://data-flair.training/blogs/python-operator/)

DataFlair introduces several classes of Python operators:

- Arithmetic
- Relational / comparison
- Assignment
- Logical
- Membership
- Identity
- Bitwise

Operators are important because they turn simple values into useful code for real-life problems.


<div class="figure"></div>

## Arithmetic Operators


```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
```

## Relational and Logical Operators


```python
score = 85
attendance = 0.90

print(score >= 90)
print(score >= 80)
print(score == 85)

passed = score >= 70 and attendance >= 0.80
honor = score >= 90 or attendance >= 0.95

print("Passed?", passed)
print("Honor?", honor)
print("Not passed?", not passed)
```

## Assignment, Membership, and Identity Operators


```python
points = 10
points += 5
# points = points + 5
print(points)

topics = ["Python", "Data Science", "AI"]
print("Python" in topics)
print("Java" not in topics)

a = None
print(a is None)
print(a is not None)
```

## Bitwise Operators Preview

Bitwise operators are less important for beginners but useful in lower-level programming, flags, and some advanced tasks.


```python
x = 5   # binary: 0101
y = 3   # binary: 0011

print("x & y:", x & y)
print("x | y:", x | y)
print("x ^ y:", x ^ y)
```

## Operator Precedence

DataFlair notes that Python follows rules of precedence. Use parentheses to make your code clearer.


```python
result1 = 2 + 3 * 4
result2 = (2 + 3) * 4

print(result1)
print(result2)
```

# Part 8. Random Numbers

**Assigned reading:** [Python Random Number — DataFlair](https://data-flair.training/blogs/python-random-number/)

DataFlair explains that Python generates **pseudorandom numbers**, which appear random but are produced by deterministic processes. The tutorial discusses `seed()`, `random()`, `randint()`, `randrange()`, `gauss()`, `choice()`, `choices()`, `sample()`, and `shuffle()`.


<div class="figure"></div>

## Basic Random Values


```python
import random

print(random.random())        # float between 0 and 1
print(random.randint(1, 10))  # integer from 1 to 10
print(random.randrange(0, 10, 2))  # random even number from 0,2,4,6,8
```

## Reproducibility with seed()

A seed lets us reproduce the same random sequence, which is useful for teaching, testing, and debugging.


```python
import random

random.seed(7)
print(random.random(), random.random(), random.random())

random.seed(7)
print(random.random(), random.random(), random.random())
```

## Random Choice, Sample, and Shuffle


```python
import random

students = ["Alice", "Ben", "Carla", "David", "Eva"]

print("One random student:", random.choice(students))
print("Two students without replacement:", random.sample(students, 2))

random.shuffle(students)
print("Shuffled list:", students)
```

## Gaussian Values Preview

DataFlair introduces `gauss(mean, standard deviation)` for real-valued distributions.


```python
import random

random.seed(7)
print(random.gauss(0, 1))
print(random.gauss(0, 1))
print(random.gauss(0, 1))
```

## Example 1: Verify AI-Generated Calculations

Suppose a GenAI tool says:

> If a student earns 45 out of 50 points, the percentage is 80%.

Let's verify with Python.


```python
earned_points = 45
total_points = 50

percentage = earned_points / total_points * 100

print("Correct percentage:", percentage)
```

## Example 2: Validate AI-Generated Text

Suppose an AI-generated project description must mention Python, data, and automation.


```python
description = '''
This project uses Python to clean data and automate weekly report generation.
'''

required_words = ["Python", "data", "automation"]

for word in required_words:
    print(word, "included?", word.lower() in description.lower())
```

## Example 3: Generate Random Practice Problems

Python can automatically generate practice problems for students.


```python
import random

a = random.randint(1, 20)
b = random.randint(1, 20)
operator = random.choice(["+", "-", "*"])

if operator == "+":
    answer = a + b
elif operator == "-":
    answer = a - b
else:
    answer = a * b

print(f"Practice problem: What is {a} {operator} {b}?")
print("Answer:", answer)
```
