A module is a Python file that contains reusable code. Python also provides many standard modules, such as `datetime`, `math`, `random`, and `re`.


## More Detail: Import Styles

Python supports several import styles:

```python
import math
from math import sqrt
import datetime as dt
from datetime import date, timedelta
```

For beginners, the clearest approach is often `import module_name`, because it makes the source of each function obvious, such as `math.sqrt()` or `datetime.date.today()`.


````python
import math

print(math.sqrt(16))
print(math.pi)
````

## Importing Specific Functions


````python
from math import sqrt

print(sqrt(25))
````

## datetime Module


````python
from datetime import datetime, date, timedelta

today = date.today()
now = datetime.now()
next_week = today + timedelta(days=7)

print("Today:", today)
print("Now:", now)
print("Next week:", next_week)
````

## Real Application Example: Assignment Deadline Reminder


````python
from datetime import date

today = date(2026, 6, 10)
deadline = date(2026, 6, 12)

days_left = (deadline - today).days

print("Days until deadline:", days_left)
````

## Hierarchical Relationship

A helpful way to understand this part is to think from small to large:

1. A **function** is a reusable block of code.
2. A **module** is usually a `.py` file that can contain functions, variables, and classes.
3. A **package** is a folder that organizes related modules.
4. A **library** is a broader reusable toolkit that may contain packages and modules.
5. A **virtual environment** is not code itself; it is an isolated workspace that manages which libraries and packages are installed for a project.


<div class="figure"><p class="figcap">Hierarchy of Python reusable code: functions, modules, packages, libraries, and virtual environments.</p></div>

## Key Differences

| Concept | Meaning | Example |
|---|---|---|
| Module | A single Python file | `math`, `random`, `my_utils.py` |
| Package | A folder containing modules | `numpy`, `pandas`, a project package |
| Library | A broader collection of reusable code | Python Standard Library, NumPy |
| Virtual environment | An isolated project environment | `.venv`, `venv`, Conda env |

# Part 3. Applied Mini Case Study: Building a Small Course Utility Module

Suppose we repeatedly need to calculate how many days remain before an assignment deadline. Instead of copying the same calculation into every script, we can place the reusable logic in a module.

A simple project could be organized like this:

```text
course_project/
├── course_utils.py
└── main.py
```

## Step 1: Put Reusable Logic in `course_utils.py`

```python
from datetime import date

def days_until(deadline):
    """Return the number of days from today until a deadline."""
    return (deadline - date.today()).days
```

The file `course_utils.py` is a **module** because it is a Python file containing reusable code.

## Step 2: Import the Function in `main.py`

```python
from datetime import date
from course_utils import days_until

deadline = date(2026, 10, 5)
remaining = days_until(deadline)

print("Days until deadline:", remaining)
```

The main program does not need to know how the calculation is implemented. It imports and reuses the function from the module.

## Why This Structure Is Useful

- The reusable calculation is written once.
- The main program stays shorter and easier to read.
- The same module can be imported by other scripts.
- If the calculation changes later, we update it in one place.

This is the basic idea behind **modular programming**.
