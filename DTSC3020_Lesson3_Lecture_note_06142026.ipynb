# Part 2. Lists

**Assigned reading:** [Python List with Examples — DataFlair](https://data-flair.training/blogs/python-list-examples/)

A list is an ordered and mutable collection. Lists are one of the most commonly used Python data structures.

Important list concepts:

- Creating lists
- Accessing items by index
- Updating items
- Adding items with `append()` or `insert()`
- Removing items with `remove()`, `pop()`, or `del`
- Sorting and reversing
- Iterating through a list


````python
# Creating a list
scores = [88, 92, 79, 95, 100]

print(scores)
print(type(scores))
print("First score:", scores[0])
print("Last score:", scores[-1])
````

## Lists Are Mutable

Mutable means the object can be changed after it is created.


````python
scores = [88, 92, 79, 95, 100]

scores[2] = 85       # update one item
scores.append(98)    # add one item at the end
scores.insert(1, 90) # insert at index 1
removed = scores.pop() # remove last item and return it

print("Updated scores:", scores)
print("Removed score:", removed)
````

## Useful List Methods


````python
topics = ["Python", "AI", "Data Science"]

topics.append("Automation")
topics.insert(1, "Programming")
topics.sort()

print(topics)
print("Number of topics:", len(topics))
print("Index of Python:", topics.index("Python"))
````

## Real Application Example 1: Gradebook Summary

A list can store assignment scores and help calculate basic summary statistics.


````python
scores = [88, 92, 79, 95, 100, 84, 91]

average = sum(scores) / len(scores)
highest = max(scores)
lowest = min(scores)

print("Scores:", scores)
print("Average:", round(average, 2))
print("Highest:", highest)
print("Lowest:", lowest)
````

## Real Application Example 2: Detecting At-Risk Scores

We can filter a list to identify students who may need support.


````python
scores = [88, 55, 92, 63, 79, 95, 100, 58]

at_risk_scores = []

for score in scores:
    if score < 70:
        at_risk_scores.append(score)

print("At-risk scores:", at_risk_scores)
````

# Part 3. Python Slice

**Assigned reading:** [Python Slice — DataFlair](https://data-flair.training/blogs/python-slice/)

Slicing lets us select part of a sequence.

General pattern:

```python
sequence[start:stop:step]
```

The stop position is not included.


<div class="figure"><p class="figcap">Indexing and slicing visualization.</p></div>

````python
students = ["Alice", "Ben", "Carla", "David", "Eva", "Frank"]

print(students[0])
print(students[-1])
print(students[1:4])
print(students[:3])
print(students[3:])
print(students[::2])
print(students[::-1])
````

## Real Application Example: Previewing and Splitting Data

Before analyzing a dataset, we often preview a small portion.


````python
survey_responses = [
    "Python is useful",
    "I like AI",
    "Data science is interesting",
    "Programming is challenging",
    "I want to learn automation",
    "Debugging takes practice",
    "I want more examples",
    "I like notebooks"
]

preview = survey_responses[:3]
last_two = survey_responses[-2:]

print("Preview:", preview)
print("Last two responses:", last_two)
````

# Part 4. Tuples

**Assigned reading:** [Python Tuple — DataFlair](https://data-flair.training/blogs/python-tuple/)

A tuple is an ordered and immutable collection. Once created, a tuple's items cannot be changed in place.

Tuples are useful for:

- Fixed records
- Coordinates
- Dates
- RGB color values
- Returning multiple values from a function
- Packing and unpacking values


````python
point = (3, 5)
rgb_color = (255, 128, 0)
date = (2026, 6, 15)

print(point)
print(rgb_color)
print(date)
print(type(point))
````

## Tuple Packing and Unpacking


````python
# Tuple packing
student_record = "Alice", "Data Science", 95

# Tuple unpacking
name, major, score = student_record

print(name)
print(major)
print(score)
````

## Returning Multiple Values from a Function


````python
def summarize_scores(scores):
    return min(scores), max(scores), sum(scores) / len(scores)

scores = [88, 92, 79, 95, 100]
lowest, highest, average = summarize_scores(scores)

print("Lowest:", lowest)
print("Highest:", highest)
print("Average:", average)
````

## Single-Item Tuple

A single-item tuple needs a comma.


````python
not_tuple = (1)
real_tuple = (1,)

print(type(not_tuple))
print(type(real_tuple))
````

# Part 5. Tuple vs List

**Assigned reading:** [Python Tuples vs Lists — DataFlair](https://data-flair.training/blogs/python-tuples-vs-lists/)

Lists and tuples are both ordered collections, but the key difference is mutability.


## List vs Tuple Summary

| Feature | List | Tuple |
|---|---|---|
| Syntax | `[ ]` | `( )` |
| Ordered | Yes | Yes |
| Mutable | Yes | No |
| Allows duplicates | Yes | Yes |
| Common use | Changing collections | Fixed records |
| Example | scores | coordinates |


````python
# List: scores can change
scores = [80, 90, 95]
scores[0] = 85
scores.append(100)
print(scores)

# Tuple: coordinate should stay fixed
coordinate = (3, 5)
print(coordinate)

# coordinate[0] = 4  # This would cause an error
````

# Part 6. Sets and Booleans

**Assigned reading:** [Python Set and Booleans — DataFlair](https://data-flair.training/blogs/python-set-and-booleans-with-examples/)

A set is an unordered collection of unique values. Sets are useful for removing duplicates and comparing groups.


````python
topics = {"Python", "AI", "Data Science", "Python", "AI"}

print(topics)
print("Python" in topics)
print("Java" in topics)
````

## Removing Duplicates

This is one of the most common beginner-friendly uses of sets.


````python
raw_tags = ["AI", "Python", "AI", "Data", "Python", "Health"]
unique_tags = set(raw_tags)

print("Original tags:", raw_tags)
print("Unique tags:", unique_tags)
````

## Set Operations

Set operations are useful in recommendation systems, survey analysis, and comparing groups.


````python
group_a = {"Alice", "Ben", "Carla"}
group_b = {"Carla", "David", "Eva"}

print("Union:", group_a | group_b)
print("Intersection:", group_a & group_b)
print("Only in A:", group_a - group_b)
print("Symmetric difference:", group_a ^ group_b)
````

## Real Application Example: Comparing Skills

Suppose two students list the tools they know. Sets can quickly show shared and unique skills.


````python
student_a_skills = {"Python", "Excel", "SQL", "AI"}
student_b_skills = {"Python", "Tableau", "SQL", "Statistics"}

shared_skills = student_a_skills & student_b_skills
a_only = student_a_skills - student_b_skills
b_only = student_b_skills - student_a_skills

print("Shared skills:", shared_skills)
print("Only student A:", a_only)
print("Only student B:", b_only)
````

## Booleans

Booleans represent truth values: `True` or `False`.


````python
score = 85
passed = score >= 70

print(passed)
print(type(passed))

if passed:
    print("The student passed.")
else:
    print("The student did not pass.")
````

# Part 7. Dictionaries

**Assigned reading:** [Python Dictionary — DataFlair](https://data-flair.training/blogs/python-dictionary/)

A dictionary stores key-value pairs. It is one of the most important data structures for real applications.

Common uses:

- Student profiles
- Configuration settings
- Word counts
- JSON-like data
- API responses
- Database-like records


<div class="figure"><p class="figcap">Dictionary key-value lookup model with cleaner spacing and larger text.</p></div>

````python
student = {
    "name": "Alice",
    "major": "Data Science",
    "score": 95,
    "passed": True
}

print(student)
print(student["name"])
print(student["score"])
````

## Updating and Adding Dictionary Values


````python
student = {
    "name": "Alice",
    "major": "Data Science",
    "score": 95
}

student["score"] = 98
student["email"] = "alice@unt.edu"

print(student)
````

## Safer Lookup with get()


````python
student = {"name": "Alice", "score": 95}

print(student.get("name"))
print(student.get("email", "No email found"))
````

## Real Application Example: Word Frequency Counter

A dictionary is often used for counting.


````python
text = "python ai python data ai python"
words = text.split()

word_counts = {}

for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print(word_counts)
````

## More Concise Counting with get()


````python
text = "python ai python data ai python"
words = text.split()

word_counts = {}

for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1

print(word_counts)
````

# Part 8. Nested Data Structures

Real data rarely uses only one structure. A common pattern is a **list of dictionaries**.


<div class="figure"><p class="figcap">A list of dictionaries is common in JSON, APIs, spreadsheets, and data science workflows.</p></div>

````python
students = [
    {"name": "Alice", "major": "Data Science", "score": 95},
    {"name": "Ben", "major": "Business", "score": 88},
    {"name": "Carla", "major": "Data Science", "score": 91},
    {"name": "David", "major": "Health Informatics", "score": 84}
]

print(students[0])
print(students[0]["name"])
print(students[2]["score"])
````

## Real Application Example: Summarize a Small Dataset


````python
students = [
    {"name": "Alice", "major": "Data Science", "score": 95},
    {"name": "Ben", "major": "Business", "score": 88},
    {"name": "Carla", "major": "Data Science", "score": 91},
    {"name": "David", "major": "Health Informatics", "score": 84}
]

scores = [student["score"] for student in students]
majors = {student["major"] for student in students}

print("Average score:", sum(scores) / len(scores))
print("Unique majors:", majors)
````

# Part 9. List Comprehension

**Assigned reading:** [Python List Comprehension — DataFlair](https://data-flair.training/blogs/python-list-comprehension/)

A list comprehension is a concise way to create a new list from an existing sequence.


<div class="figure"><p class="figcap">List comprehension transforms and filters data concisely.</p></div>

## Traditional Loop vs List Comprehension


````python
numbers = [1, 2, 3, 4, 5]

# Traditional loop
squares_loop = []
for n in numbers:
    squares_loop.append(n * n)

# List comprehension
squares_comp = [n * n for n in numbers]

print(squares_loop)
print(squares_comp)
````

## Filtering with List Comprehension


````python
scores = [55, 72, 88, 40, 95, 67]

passing_scores = [score for score in scores if score >= 70]

print(passing_scores)
````

## Real Application Example: Clean Survey Responses


````python
responses = [" Python ", "AI", " data science ", "PYTHON", "ai ", ""]

cleaned = [response.strip().lower() for response in responses if response.strip() != ""]

print(cleaned)
````

# Part 10. Generative AI-Era Examples

## Example 1: Clean AI-Generated Topic Lists

Suppose a GenAI tool generates a list of course topics, but there are duplicates and inconsistent capitalization.


````python
ai_topics = ["Python", "AI", "python", "Data Science", "AI", "data science", "  Python  "]

clean_topics = [topic.strip().lower() for topic in ai_topics]
unique_topics = set(clean_topics)

print("Cleaned topics:", clean_topics)
print("Unique topics:", unique_topics)
````

## Example 2: Convert Semi-Structured AI Output into a Dictionary

AI may produce text, but Python can help structure it.


````python
student_text = "name=Alice;major=Data Science;score=95"

parts = student_text.split(";")
student = {}

for part in parts:
    key, value = part.split("=")
    student[key] = value

print(student)
print(student["major"])
````

## Example 3: Validate AI-Generated Records

Suppose an AI system generates student records. We can check whether each record contains required keys.


````python
records = [
    {"name": "Alice", "major": "Data Science", "score": 95},
    {"name": "Ben", "score": 88},
    {"name": "Carla", "major": "Data Science", "score": 91}
]

required_keys = {"name", "major", "score"}

for record in records:
    missing_keys = required_keys - set(record.keys())
    if missing_keys:
        print("Incomplete record:", record, "| Missing:", missing_keys)
    else:
        print("Complete record:", record)
````

## Example 4: Simple Recommendation Lookup

A dictionary can power a simple recommendation system.


````python
resource_recommendations = {
    "list": "Use lists for ordered, changeable data.",
    "tuple": "Use tuples for fixed records.",
    "set": "Use sets to remove duplicates.",
    "dictionary": "Use dictionaries for key-value lookup."
}

topic = "dictionary"
print(resource_recommendations.get(topic, "Topic not found."))
````

# Part 11. Applied Mini Case Study: Course Feedback Analyzer

In this mini case, we use several data structures together to analyze student feedback.


````python
feedback = [
    "Python is useful",
    "I like Python examples",
    "AI examples are interesting",
    "I need more practice",
    "Python and AI are useful",
    "Examples help me learn"
]

# Step 1: Combine all feedback into one text
all_text = " ".join(feedback).lower()

# Step 2: Split into words
words = all_text.split()

# Step 3: Count words using a dictionary
word_counts = {}

for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1

# Step 4: Extract unique words using a set
unique_words = set(words)

print("Word counts:", word_counts)
print("Unique words:", unique_words)
print("Number of unique words:", len(unique_words))
````

## Mini Case Reflection

This small example used:

- List: store multiple feedback responses
- String: combine and process text
- Dictionary: count words
- Set: identify unique words
- Loop: process each word
